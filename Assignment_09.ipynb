```python
gamma = 0.9

# Solve the Bellman equations algebraically
V0 = 71.819 / 0.9271
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1
Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V0

print(f"V(S0) = {V0:.2f}")
print(f"V(S1) = {V1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q(S1, Up) = {Q_S1_Up:.2f}")

# Verify the Bellman expectation equations
check_V0 = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
check_V1 = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print(f"Verified V(S0) = {check_V0:.2f}")
print(f"Verified V(S1) = {check_V1:.2f}")
```

```python
gamma = 0.9
theta = 0.000001

# Initialize state values
V0 = 0.0
V1 = 0.0
iterations = 0

while True:
    # Calculate new state values
    new_V0 = 0.1 * (-1 + gamma * 0) + \
             0.9 * (-1 + gamma * V1)

    new_V1 = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * V0)

    # Calculate maximum change
    delta = max(abs(new_V0 - V0),
                abs(new_V1 - V1))

    # Update state values
    V0 = new_V0
    V1 = new_V1
    iterations += 1

    # Check convergence
    if delta < theta:
        break

# Calculate action values after convergence
Q_S0_Down = 0.1 * (-1 + gamma * 0) + \
            0.9 * (-1 + gamma * V1)

Q_S0_Right = -1 + gamma * 0


Q_S1_Up = -1 + gamma * V0

# Display results
print("Iterations:", iterations)
print(f"V(S0) = {V0:.6f}")
print(f"V(S1) = {V1:.6f}")
print(f"Q(S0, Down) = {Q_S0_Down:.6f}")
print(f"Q(S0, Right) = {Q_S0_Right:.6f}")
print(f"Q(S1, Right) = {Q_S1_Right:.6f}")
print(f"Q(S1, Up) = {Q_S1_Up:.6f}")

# Verify policy consistency
print("Policy check S0:",
      abs(V0 - Q_S0_Down) < theta)

print("Policy check S1:",
      abs(V1 - Q_S1_Right) < theta)
```
