```python
# Multi-Armed Bandit using Greedy Algorithm
# Q(a) is calculated using the Arithmetic Mean

n = int(input("Enter number of machines: "))
machines = {}
q_values = {}
for i in range(n):
    name = input(f"\nEnter machine name : ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))
    machines[name] = rewards
    q = sum(rewards) / len(rewards)
    q_values[name] = q
print("\n--- Estimated Q Values ---")
for name in machines:
    print(f"\nMachine {name}")
    print(f"Rewards: {machines[name]}")
    print(f"Q({name}) = {q_values[name]:.2f}")
selected_machine = max(q_values, key=q_values.get)
print(f"\nGreedy Selected Machine: {selected_machine}")
```

```python
# Multi-Armed Bandit using Incremental Update Method

n = int(input("Enter number of machines: "))
machines = {}
Q = {}
N = {}
for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards separated by spaces: ").split()))
    machines[name] = rewards
    Q[name] = 0
    N[name] = 0
for name in machines:
    for reward in machines[name]:
        N[name] += 1
        Q[name] = Q[name] + (reward - Q[name]) / N[name]
print("\n--- Estimated Q Values ---")
for name in machines:
    print(f"Machine {name}")
    print(f"Rewards: {machines[name]}")
    print(f"Q({name}) = {Q[name]:.1f}")
    print()
selected_machine = max(Q, key=Q.get)
print(f"Greedy Selected Machine: {selected_machine}")
```
