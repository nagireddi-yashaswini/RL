```python
# Multi-Armed Bandit - Greedy Algorithm
# Arithmetic Mean Method

# Number of machines
n = int(input("Enter number of machines: "))

machines = {}

# Take rewards for each machine
for i in range(n):
    name = input(f"\nEnter name of machine {i + 1}: ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))
    machines[name] = rewards

print("\n--- Estimated Q Values ---")

q_values = {}

# Calculate arithmetic mean
for name, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[name] = q

    print(f"Machine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q:.2f}")

# Greedy selection
selected_machine = max(q_values, key=q_values.get)

print("\nGreedy Selected Machine:", selected_machine)
```

```python
# Multi-Armed Bandit - Greedy Algorithm
# Incremental Update Method

# Number of machines
n = int(input("Enter number of machines: "))

machines = {}
Q = {}
N = {}

# Take machine names and rewards
for i in range(n):
    name = input(f"\nEnter name of machine {i + 1}: ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))

    machines[name] = rewards
    Q[name] = 0
    N[name] = 0

print("\n--- Incremental Updates ---")

# Process every reward
for name, rewards in machines.items():

    print(f"\nMachine {name}")

    for reward in rewards:

        # Increase number of plays
        N[name] += 1

        # Incremental update
        Q[name] = Q[name] + (reward - Q[name]) / N[name]

        print(f"Reward = {reward}")
        print(f"N({name}) = {N[name]}")
        print(f"Q({name}) = {Q[name]:.2f}")

# Display final Q values
print("\n--- Final Q Values ---")

for name in machines:
    print(f"Machine {name}")
    print(f"Rewards: {machines[name]}")
    print(f"Q({name}) = {Q[name]:.2f}")

# Greedy selection
selected_machine = max(Q, key=Q.get)

print("\nGreedy Selected Machine:", selected_machine)
```
