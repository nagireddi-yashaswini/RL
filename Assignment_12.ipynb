```python
# Value Iteration in Deterministic Grid World

gamma = 0.9
theta = 0.001

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

# Transition model: next state and reward
transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 100)
    }
}

# Initialize state values
V = {"S0": 0, "S1": 0, "G": 0}

print("Iteration-wise State Values")
print("Iteration | V(S0) | V(S1)")

iteration = 0

while True:
    delta = 0
    new_V = V.copy()

    for state in ["S0", "S1"]:
        q_values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)
        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    iteration += 1

    print(f"{iteration:9} | {V['S0']:6.2f} | {V['S1']:6.2f}")

    if delta < theta:
        break

# Calculate final Q-values and optimal policy
print("\nFinal Q-values and Optimal Policy")

for state in ["S0", "S1"]:
    q_values = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        q = reward + gamma * V[next_state]
        q_values[action] = q

    best_action = max(q_values, key=q_values.get)

    print(f"\nState: {state}")
    for action, q in q_values.items():
        print(f"{action:5} : {q:.2f}")

    print("Optimal Action:", best_action)
    print("Optimal Value:", V[state])

print("\nTerminal State Value V(G):", V["G"])
```
